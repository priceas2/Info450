```python
wrestlers=['John Cena', 'The Rock', 'Stone Cold', 'Sin Cara']
matches= [1500,800,1200,1100]
```

```python
wrestlers_series=pd.Series(wrestlers)
matches_series=pd.Series(matches)
```

```python
wrestlers_series
```

```python
df=pd.DataFrame(columns=(['Wrestlers', 'Matches']))
```

```python
df['Wrestlers']=wrestlers_series
df['Matches']=matches_series
```

```python
df
```

```python
df['Wins']=[12,45,50,5]
df['Win Ratio']=df['Wins']/df['Matches']
```

```python
df
```

```python
df.plot(x='Wrestlers', y='Matches', kind='bar')
```
