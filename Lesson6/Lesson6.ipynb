# Logical Operators

```python
age=25
income=50000
#Want to make sure they are an adult making 40k a year
if age>18 and income>40000:
  print("Congrats on the loan!")
```

```python
age=17
income=50000
#Want to make sure they are going to pay me back
if (age>18 and income>40000) or income>4000000:
  print("Congrats on the loan!")
```

```python
hours_worked=40
performance_score=85
tenure=2 #Years with company
if (hours_worked>=40 and performance_score >80) or tenure>=2:
  print("Promoted!")
else:
  print("No promotion")
```

# Nested conditions

```python
x=5
y=4
if x==y:
  print("x and y are the same")
else:
  if x<y:
    print("x is smaller")
  else:
    print("y is smaller")
  print("outside of if statement")
```
